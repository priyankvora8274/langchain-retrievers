```python
import os

os.environ["OPENAI_API_KEY"] = ""
```

```python
from langchain_openai import ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_core.documents import Document
from langchain_classic.retrievers.multi_query import MultiQueryRetriever
```

```python

doc1 = Document(
    page_content="""Python is widely used for developing machine learning applications.
It provides libraries such as Scikit-learn, TensorFlow, and PyTorch."""
)

doc2 = Document(
    page_content="""Artificial intelligence enables computers to perform tasks that normally require human intelligence.
AI systems can recognize patterns, make predictions, and automate decisions."""
)

doc3 = Document(
    page_content="""Machine learning algorithms learn patterns from historical data to make predictions.
Supervised learning uses labeled datasets to train predictive models."""
)

doc4 = Document(
    page_content="""Deep learning uses artificial neural networks with multiple layers to process complex data.
It is commonly used in image recognition, speech processing, and natural language processing."""
)

doc5 = Document(
    page_content="""Data scientists use Python libraries like Pandas and NumPy for data preprocessing.
Clean and structured datasets help improve the performance of predictive models."""
)

doc6 = Document(
    page_content="""AWS SageMaker is a cloud service for building, training, and deploying machine learning models.
It provides managed infrastructure for machine learning workflows."""
)

doc7 = Document(
    page_content="""Natural language processing allows computers to understand and generate human language.
NLP techniques are used in chatbots, text classification, and sentiment analysis."""
)

doc8 = Document(
    page_content="""Feature engineering involves selecting and transforming raw data into useful input variables.
Well-designed features can significantly improve machine learning model accuracy."""
)

doc9 = Document(
    page_content="""SQL is used to retrieve, filter, and aggregate data from relational databases.
Data engineers use SQL queries to prepare datasets for analytics and reporting."""
)

doc10 = Document(
    page_content="""Docker packages applications and their dependencies into portable containers.
Developers use Docker to deploy software consistently across different environments."""
)

# Store all documents in a list
documents = [
    doc1, doc2, doc3, doc4, doc5,
    doc6, doc7, doc8, doc9, doc10
]
```

```python
vector_db = FAISS.from_documents(
    documents = documents,
    embedding = OpenAIEmbeddings()
)
```

```python
simple_retriever = vector_db.as_retriever(search_type="similarity",kwargs={"k":5})
```

```python
mqr_retriever  = MultiQueryRetriever.from_llm(
    retriever = vector_db.as_retriever(kwargs={"k":5}),
    llm=ChatOpenAI(model = 'gpt-5.2'),
)
```

```python
docs1 = simple_retriever.invoke("How can I make my Python API faster and handle more users??")
```

```python
docs2 = mqr_retriever.invoke("How can I make my Python API faster and handle more users?")
```

```python
for i,doc in enumerate(docs1):
  print(f"Result {i} : {doc.page_content}")

print("*"*100)

for i,doc in enumerate(docs2):
  print(f"Result {i} : {doc.page_content}")
```
