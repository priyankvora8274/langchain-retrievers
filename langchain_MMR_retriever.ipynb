```python
import os

os.environ["OPENAI_API_KEY"] = "API_key"
```

```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
```

```python

doc1 = Document(
    page_content="""Python is widely used for machine learning and artificial intelligence.
It provides libraries like TensorFlow and Scikit-learn for building AI models."""
)

doc2 = Document(
    page_content="""Python is a popular programming language for developing machine learning applications.
Developers use Python libraries to train and evaluate artificial intelligence models."""
)

doc3 = Document(
    page_content="""Machine learning engineers prefer Python because of its simple syntax and powerful libraries.
Python supports deep learning, data analysis, and AI model development."""
)

doc4 = Document(
    page_content="""AWS provides cloud computing services such as EC2, Lambda, and S3.
Companies use AWS to deploy applications and manage cloud infrastructure."""
)

doc5 = Document(
    page_content="""SQL is used to retrieve and manipulate data stored in relational databases.
Database developers write SQL queries to filter, join, and aggregate records."""
)

# Store all documents in a list
documents = [doc1, doc2, doc3, doc4, doc5]
```

```python
vectordb = FAISS.from_documents(
    documents = documents,
    embedding=OpenAIEmbeddings()
)
```

```python
retriever = vectordb.as_retriever(
    search_type="mmr",  # Maximal Marginal Relevance: balances relevance and diversity
    search_kwargs={
        "k": 3,          # Number of final documents to retrieve (top 3 results)
        "lambda_mult": 0.5  # Controls balance between relevance and diversity (0 to 1)
                            # 0.0 = Maximum diversity (avoids similar/redundant documents)
                            # 0.5 = Balanced relevance and diversity
                            # 1.0 = Maximum relevance (prioritizes documents similar to query)
    }
)
```

```python
docs = retriever.invoke("How is Python used in machine learning and software development?")
```

```python
for i,doc in enumerate(docs):
  print(f"Result {i} : {doc.page_content}")
```
